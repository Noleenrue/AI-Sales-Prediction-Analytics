SALES DISTRIBUTION GRAPH

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Sales"], bins=50, color='skyblue', edgecolor='black')

plt.title("Distribution of Sales", fontdict={'fontsize': 16, 'fontweight': 'bold'})
plt.xlabel("Sales", fontdict={"fontsize": 14})
plt.ylabel("Frequency", fontdict={"fontsize": 14})
plt.show()
```

SALES BY CATEGORY VISUALIZATION

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

category_sales
```

```python
category_sales.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)

plt.show()
```

SALES BY PROFIT VISUALIZATION

```python
category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

category_profit
```

```python
category_profit.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.xticks(rotation=0)

plt.show()
```

INSIGHT
The category with the highest sales isn't necessarily the category with the highest profit
